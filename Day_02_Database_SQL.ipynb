```python
df = pd.read_csv('/content/student_performance.csv') # Path might vary
```

```python
import sqlite3

# 1. Connect to your database
conn = sqlite3.connect('college.db')

# 2. Push the new table from your friend's file
df.to_sql('students', conn, if_exists='replace', index=False)

# 3. Check if it worked
cursor = conn.cursor()
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
print("Confirmed Tables:", cursor.fetchall())

conn.close()
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('college.db')

# Query 1: See the first 10 students
print("--- First 10 Students ---")
query1 = "SELECT * FROM students LIMIT 10"
print(pd.read_sql_query(query1, conn))

# Query 2: Get basic statistics (count of students)
print("\n--- Total Number of Students ---")
query2 = "SELECT COUNT(*) as total_students FROM students"
print(pd.read_sql_query(query2, conn))

conn.close()
```
