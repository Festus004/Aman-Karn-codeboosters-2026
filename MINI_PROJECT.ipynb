```python
# CELL 1: Define raw messy data
messy_invoices = [
    "Invoice #INV-9901 from Alpha Tech Solutions on 2026-05-12. Total due is Rs. 12,500 for Web Hosting Service.",
    "Billed by ByteCraft Labs: order 4412 on Feb 14, 2026, amount charging Rs. 8,200 for UI Consultation.",
    "CloudNexus Inc issued invoice 2026-883 on 03/22/2026. Rs. 45,000 for Cloud Migration.",
    "Invoice from DevOps Pro Co. Number: 1004. Billed on May 02, 2026. Costing Rs. 18,500 for Docker Setup.",
    "Apex Analytics charged Rs. 31,000 via Invoice #APX-77 on 11th January 2026 for Data Dashboard Development."
]

print(f"Loaded {len(messy_invoices)} messy invoices successfully.")
```

```python
# CELL 2: Create the LLM Processor using a Strong Prompt
import json

def ask_llm_for_cleaning(invoice_text):
    """
    Simulates a strong prompt LLM call that extracts fields into structured JSON.
    """
    # The Strong Prompt Blueprint
    strong_prompt = f"""
    You are an automated ETL data processing assistant.
    Extract the invoice details from the following messy text into a strict JSON object.

    Messy Text: "{invoice_text}"

    Required Keys:
    - "invoice_id" (Extract alphanumeric id or number)
    - "vendor" (Company name providing the bill)
    - "date" (Standardize to YYYY-MM-DD format)
    - "amount_inr" (Extract numbers only as an integer)
    - "description" (The service description)

    Rules: Return ONLY raw JSON. No markdown blocks, no greeting text.
    """

    # Simulating the exact structural output a strong prompt model produces
    # (In your real notebook, you would replace this return with: return ask_llm(strong_prompt, temperature=0.0))
    if "Alpha Tech" in invoice_text:
        return '{"invoice_id": "INV-9901", "vendor": "Alpha Tech Solutions", "date": "2026-05-12", "amount_inr": 12500, "description": "Web Hosting Service"}'
    elif "ByteCraft" in invoice_text:
        return '{"invoice_id": "4412", "vendor": "ByteCraft Labs", "date": "2026-02-14", "amount_inr": 8200, "description": "UI Consultation"}'
    elif "CloudNexus" in invoice_text:
        return '{"invoice_id": "2026-883", "vendor": "CloudNexus Inc", "date": "2026-03-22", "amount_inr": 45000, "description": "Cloud Migration"}'
    elif "DevOps Pro" in invoice_text:
        return '{"invoice_id": "1004", "vendor": "DevOps Pro Co.", "date": "2026-05-02", "amount_inr": 18500, "description": "Docker Setup"}'
    elif "Apex Analytics" in invoice_text:
        return '{"invoice_id": "APX-77", "vendor": "Apex Analytics", "date": "2026-01-11", "amount_inr": 31000, "description": "Data Dashboard Development"}'

    return '{}'
```

```python
# CELL 3: Process text into a collection of JSON records
cleaned_records = []

print("Starting GenAI ETL Pipeline Processing...\n")

for index, raw_text in enumerate(messy_invoices):
    # 1. Run text through LLM
    raw_llm_output = ask_llm_for_cleaning(raw_text)

    try:
        # 2. Parse string output into a python dictionary
        parsed_json = json.loads(raw_llm_output.strip())
        cleaned_records.append(parsed_json)
        print(f"✔ Successfully processed Invoice {index + 1}")
    except json.JSONDecodeError:
        print(f"✖ Failed to parse Invoice {index + 1} due to unformatted output.")

print("\nAll records processed.")
```

```python
# CELL 4: Convert JSON Collection into a Pandas DataFrame
import pandas as pd

# Load the structured python dictionaries directly into a DataFrame
df = pd.DataFrame(cleaned_records)

# Verify the structural types are properly configured
df['amount_inr'] = df['amount_inr'].astype(int)
df['date'] = pd.to_datetime(df['date'])

# Display the clean DataFrame
print("--- CLEANED STRUCTURED DATAFRAME ---")
print(df)
```

```python
# CELL 5: Run metric computations and data analysis
print("=== INVOICE METRICS ANALYSIS ===\n")

# 1. Total Expenditure Calculation
total_spend = df['amount_inr'].sum()
print(f"Total Business Spend: Rs. {total_spend:,}")

# 2. Average Invoice Cost
avg_spend = df['amount_inr'].mean()
print(f"Average Invoice Amount: Rs. {avg_spend:,}")

# 3. Maximum Expense Identification
highest_expense = df.loc[df['amount_inr'].idxmax()]
print(f"Highest Expense Vendor: {highest_expense['vendor']} (Rs. {highest_expense['amount_inr']:,})")
```
