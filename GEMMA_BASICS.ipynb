```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# Define the official path repository for Gemma
model_id = "google/gemma-1.1-2b-it"

print("Downloading and loading the Gemma Tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(model_id)

print("Downloading and loading the Gemma Model weights onto the T4 GPU...")
# We load using bfloat16 to optimize memory usage without losing intelligence
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    torch_dtype=torch.bfloat16,
    device_map="auto"
)

print("\nGemma is successfully loaded and ready for inference!")
```

```python
print("Preparing the structured prompt environment...")

# 1. Define your instruction message using standard role-based dictionaries
messages = [
    {
        "role": "user",
        "content": "Explain the primary benefits of containerizing an enterprise application using Docker."
    }
]

# 2. Apply Gemma's official chat template format
prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

# 3. Convert raw text prompt into mathematical token numbers and move them to the T4 GPU
inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

print("Generating response from Gemma (this takes just a few seconds)...")

# 4. Run the generation math engine
outputs = model.generate(
    **inputs,
    max_new_tokens=250,  # Limits the response length to around 250 words
    temperature=0.7,     # Controls creativity (0.7 is perfect for balanced technical answers)
    do_sample=True       # Enables sampling optimization parameters
)

# 5. Decode the numerical tokens back into human-readable text
response = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("\n" + "="*50)
print("GEMMA AI OUTPUT:")
print("="*50)
print(response)
print("="*50)
```

```python
def ask_gemma(user_question: str, creativity: float = 0.7):
    """
    Reusable inference function to chat with our loaded Gemma model.
    """
    # 1. Format the text inside Gemma's instruction structure
    chat_history = [{"role": "user", "content": user_question}]
    formatted_prompt = tokenizer.apply_chat_template(chat_history, tokenize=False, add_generation_prompt=True)

    # 2. Tokenize and push vectors to the GPU
    model_inputs = tokenizer(formatted_prompt, return_tensors="pt").to("cuda")

    # 3. Generate tokens using the parameterized temperature dial
    generated_tokens = model.generate(
        **model_inputs,
        max_new_tokens=300,
        temperature=creativity,
        do_sample=True
    )

    # 4. Decode the result back into human-readable text
    cleaned_output = tokenizer.decode(generated_tokens[0], skip_special_tokens=True)

    print("\n" + "-"*40)
    print(f"User Query: {user_question}")
    print("-"*40)
    print(cleaned_output)
    print("-"*40 + "\n")

# ========================================================
# TEST RUNS: Test the system with a couple of different topics
# ========================================================
ask_gemma("What is the difference between a Relational Database and a NoSQL Database?")
ask_gemma("Write a simple Python script to calculate the factorial of a number.")
```
